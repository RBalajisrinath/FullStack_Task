# SaaSquatch v2 — API demo walkthrough
Run backend first: `python -m uvicorn backend.main:app --port 8000`. Then run cells.

```python
import urllib.request, json
BASE='http://localhost:8000'
def get(p):
    return json.load(urllib.request.urlopen(BASE+p))
print(get('/api/health'))
```

```python
s=get('/api/stats'); print(s)
# distribution proves prioritization value: A-tier is the callable 40%
```

```python
top=get('/api/leads?min_score=88&verify=true&page_size=3')
for x in top['items']: print(x['score'], x['tier'], x['company'], x['email'], x['verify']['email_status'])
print('total A-tier:', top['total'], '| deduped:', top['removed_duplicates'])
```

```python
# export CRM-ready CSV
import pathlib
urllib.request.urlretrieve(BASE+'/api/export?min_score=73', 'v2_priority_leads.csv')
print('saved v2_priority_leads.csv')
```
